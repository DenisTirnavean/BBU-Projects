```python
from scipy.stats import norm
from matplotlib.pyplot import hist,xticks, grid, show, plot
from numpy import linspace, mean, std
m = 165
sigma = 10
data = norm.rvs(loc=m, scale=sigma,size = 1000)
hist(data, bins = 16, range=(130,210),density = True)
xticks(range(130,210,5), fontsize = 8)
x = linspace(130,210,1000)
plot(x,norm.pdf(x,loc = m, scale = sigma), color = 'red')
grid()
show()
```

```python
norm.pdf(x,m,sigma) #fX(x) functia de densitate
print(mean(data)) #E(X) = m
print(std(data))  #Sigma(X) = sigma

print("Probabilitate estimata: ",end="")
print(mean([160<x<170 for x in data]))

prob  = norm.cdf(170,loc = m, scale = sigma) - norm.cdf(160,loc = m, scale = sigma)
print("Probabilate teoretica: "+str(prob))
```

```python
from scipy.stats import expon, uniform
from numpy import mean,std, multiply

from random import random

def chose_printer():
  choice = uniform.rvs()
  if(choice<=0.4):
    return "I1"
  return "I2"

def get_t():
  printer = chose_printer()
  return expon.rvs(loc = 0, scale =1/5) if printer == "I1" else uniform.rvs(loc = 4, scale = 6-4)

data = [get_t() for _ in range(1000)]
print("Media estimata: "+ str(mean(data)))
print("Probabilitate estimata ca sa se printeze in <5 secs: "+ str(mean([x<5 for x in data])))
print("Probabilitate teoretica ca sa se printeze in <5 secs: "+ str(expon.cdf(5,loc = 0,scale = 5)*0.4 + uniform.cdf(5, loc = 4, scale = 2)*0.6))
```
