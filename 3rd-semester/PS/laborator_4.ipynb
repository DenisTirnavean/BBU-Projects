```python
from scipy.stats import bernoulli, binom
from matplotlib.pyplot import bar, show, hist, grid, legend, xticks
bernoulli.rvs(p=0.5, size=10)

def rand_walk(num, p):
  poz =0
  poz_viz = [0]
  for _ in range(num):
    is_right = bernoulli.rvs(p, size=1)
    if(is_right):
      poz += 1
    else:
      poz -= 1
    poz_viz.append(poz)
  return poz_viz
p =0.5
num = int(input("Numar pasi = "))
print("Traseul este: ", rand_walk(num,p))
```
