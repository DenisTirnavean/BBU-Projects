```python
#asezare 5 persoane pe 12 scaune cu space intre ele
from copy import deepcopy
```

```python
pozitii = list(combinations(range(8),5))
perm_pers = list(permutations("12345"))
pers = "12345"
toate_asezarile = []
for pozitie in pozitii:
  scaune = list("_"*8)
  for pers in perm_pers:
    for i in range(5):
      scaune[pozitie[i]] = pers[i]
    toate_asezarile.append(scaune)
len(toate_asezarile)
toate_asezarile[666]
solutie = []
#partea de inserturi
for scaune in toate_asezarile:
  scaune_noi = deepcopy(scaune)
  for i in range(len(scaune)):
    if scaune[i]!='_':
      scaune_noi.insert(i+1,"_")
  solutie.append(scaune_noi)
solutie
```
