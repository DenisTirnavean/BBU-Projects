```python
def create_rmd(x: list, p: list)-> int:
  rando = uniform.rvs()
  intervals=[[0,p[0]]]
  for prob in range(1,len(p)):
    intervals.append([intervals[prob-1][1],intervals[prob-1][1]+p[prob]])
  for it in intervals:
    if rando >= it[0] and rando < it[1]:
      return x[intervals.index(it)]
  return 'nimic'

#create_rmd([1,2,3,4,5,6],[0.1666666666666667,0.1666666666666667,0.1666666666666667,0.1666666666666667,0.1666666666666667,0.1666666666666667])
print([create_rmd([1,2,3,4,5,6],[1/6,1/6,1/6,1/6,1/6,1/6]) for _ in range(100)])

```

```python
from random import randrange
from matplotlib.pyplot import bar, hist, grid, show, legend


#x= [1,2,3,4,5,6]
x= [-1,1,3]
#p=[1/6,1/6,1/6,1/6,1/6,1/6]
p= [0.2, 0.5, 0.3]
data = [create_rmd(x,p) for _ in range(1000)];
bin_edges = [k-0.5 for k in range(min(x),max(x)+2)]
hist(data, bin_edges , density = True, rwidth = 0.85 ,color = 'chartreuse', edgecolor = 'black',alpha = 1, label = 'frecvente relative')

bar(x, p, width = 0.85, color = '#b54374', edgecolor = 'black',alpha= 0.5, label = 'probabilitati')
```

```python
from math import log
def return_formula(a):
  y = uniform.rvs()
  x = -1/a*log(1-y)
  return x
print(return_formula(1/10))
```
