```python
def simulari_zile_nastere(n: int, nr_sim = 1000): #n numar de persoane
  contor = 0
  for _ in range(nr_sim):
    zile_nastere  = [randint(1,365) for _ in range(n)]
    if len(set(zile_nastere)) != n:
      contor += 1
  return (contor/nr_sim)
simulari_zile_nastere(30)
```

```python
from matplotlib.pyplot import axis,plot,show
from random import random
from math import dist
```

```python
axis('square')
axis((0,1,0,1))
for _ in range(1000):
  F=[random(),random()]
  plot(F[0],F[1],'ro')
show()
```

```python
axis('square')
counter = 0
contor2 = 0
axis((0,1,0,1))
nr_iterations = 1000

for _ in range(nr_iterations):
  F=[random(),random()]
  if dist(F,(0.5,0.5)) <= 0.5:
    if dist(F,(0.5,0.5)) == 0.5: contor2+=1
    plot(F[0],F[1],'bo')
    counter+=1
  else:
    plot(F[0],F[1],'ro')
print(contor2)
print(counter/nr_iterations)
show()
```

```python
axis('square')
counter = 0

axis((0,1,0,1))
nr_iterations = 1000

for _ in range(nr_iterations):
  F=[random(),random()]
  x = (F[0],F[1])
  #caz centru
  dist_center = dist(F,(0.5,0.5))
  dist_A = dist(F,(0,0))
  dist_B = dist(F,(1,0))
  dist_C = dist(F,(1,1))
  dist_D = dist(F,(0,1))
  if dist_center <dist_A and dist_center <dist_B and dist_center <dist_C and dist_center <dist_D:
    plot(F[0],F[1],'bo')
    counter+=1
  else:
    plot(F[0],F[1],'ro')

print(counter/nr_iterations)
show()
```

```python
axis('square')
counter = 0
contor2 = 0
axis((0,1,0,1))
nr_iterations = 100000
A = (0,0)
B = (1,0)
C = (1,1)
D = (0,1)
for _ in range(nr_iterations):
  F=[random(),random()]
  af = dist(F,A)
  bf = dist(F,B)
  cf = dist(F,C)
  fd = dist(F,D)

  contor_asc = 0
  if(af**2 + bf**2 >1): #A cu B
    contor_asc+=1
  if(bf**2 + cf**2 >1): #B cu C
    contor_asc+=1
  if(cf**2 + fd**2 >1): #C cu D
    contor_asc+=1
  if (af**2 + fd**2 >1): #A cu D
    contor_asc+=1
  if(contor_asc ==2):
    plot(F[0],F[1],'bo')

  else:
    plot(F[0],F[1],'ro')

print(counter/nr_iterations)
show()
```
